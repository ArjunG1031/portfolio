```python
%%html
<!-- UI_RUNNER: Toggles Homework - Toggle Control Panel -->

<div class="ocs__card">
  <h3>Control Panel</h3>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="show-message" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show Message</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="show-tip" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Show Study Tip</span>
  </label>

  <label class="ocs__toggle ocs__toggle--green">
    <input id="notifications" class="ocs__toggle-input" type="checkbox">
    <span class="ocs__toggle-track" aria-hidden="true"></span>
    <span class="ocs__toggle-label">Notifications</span>
  </label>

  <p id="message" hidden>Hello! This message is controlled by the first toggle.</p>
  <p id="tip" hidden>Tip: Predict what your code will output before you run it.</p>
  <p id="notif-status">Notifications are OFF.</p>

  <output id="toggle-count" aria-live="polite">Toggles on: 0 of 3</output>
</div>

<script>
// I learned that .checked lets JavaScript tell whether a toggle is currently on or off.
const showMessage = document.getElementById("show-message");
const showTip = document.getElementById("show-tip");
const notifications = document.getElementById("notifications");

const message = document.getElementById("message");
const tip = document.getElementById("tip");
const notifStatus = document.getElementById("notif-status");
const toggleCount = document.getElementById("toggle-count");

const toggles = [showMessage, showTip, notifications];

// The toggle is a hidden checkbox, so its state is stored in .checked and the
// "change" event fires every time it is flipped.
function updatePage() {
  message.hidden = !showMessage.checked;   // hidden = true when the toggle is off
  tip.hidden = !showTip.checked;
  notifStatus.textContent = "Notifications are " + (notifications.checked ? "ON." : "OFF.");

  // Count how many toggles are on by checking each one's .checked value
  let count = 0;
  for (const t of toggles) {
    if (t.checked) {
      count++;
    }
  }
  toggleCount.textContent = "Toggles on: " + count + " of " + toggles.length;
}

// This is selection from AP CSP: the page changes based on whether a condition is true or false.
for (const t of toggles) {
  t.addEventListener("change", updatePage);
}

updatePage();
</script>
```
